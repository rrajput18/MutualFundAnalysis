# Day 1: Data Ingestion & Raw Profiling

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Define relative paths
raw_dir = Path('../data/raw')
print(f"Ingestion directory target: {raw_dir.resolve()}")
```

## 1. Directory Scan & Size Profiling

```python
raw_files = sorted(list(raw_dir.glob('*.csv')))
file_details = []
for f in raw_files:
    try:
        df = pd.read_csv(f)
        file_details.append({
            "File Name": f.name,
            "Size (KB)": round(f.stat().st_size / 1024, 2),
            "Rows": df.shape[0],
            "Columns": df.shape[1]
        })
    except Exception as e:
        print(f"Error loading {f.name}: {e}")

details_df = pd.DataFrame(file_details)
details_df
```

## 2. Ingesting & Profiling Key Datasets

```python
# Fund Master profile
fm = pd.read_csv(raw_dir / '01_fund_master.csv')
print(f"Fund Master Shape: {fm.shape}")
print("\n--- Columns and Types ---")
fm.info()
print("\n--- Sample Data ---")
fm.head(3)
```

```python
# NAV History profile
nav = pd.read_csv(raw_dir / '02_nav_history.csv')
print(f"NAV History Shape: {nav.shape}")
print("\n--- Columns and Types ---")
nav.info()
print("\n--- Sample Data ---")
nav.head(3)
```

```python
# Transactions profile
tx = pd.read_csv(raw_dir / '08_investor_transactions.csv')
print(f"Transactions Shape: {tx.shape}")
print("\n--- Columns and Types ---")
tx.info()
print("\n--- Sample Data ---")
tx.head(3)
```

## 3. Reference Mapping Validation

```python
master_amfi = set(fm['amfi_code'].unique())
nav_amfi = set(nav['amfi_code'].unique())
tx_amfi = set(tx['amfi_code'].unique())

print(f"Unique AMFI codes in Fund Master: {len(master_amfi)}")
print(f"Unique AMFI codes in NAV History: {len(nav_amfi)}")
print(f"Unique AMFI codes in Transactions: {len(tx_amfi)}")

# Detect missing references
nav_orphans = nav_amfi - master_amfi
tx_orphans = tx_amfi - master_amfi

print(f"\nOrphaned AMFI codes in NAV History (not in master): {nav_orphans}")
print(f"Orphaned AMFI codes in Transactions (not in master): {tx_orphans}")
if not nav_orphans and not tx_orphans:
    print("Referential integrity checks passed! All AMFI codes map correctly.")
```
